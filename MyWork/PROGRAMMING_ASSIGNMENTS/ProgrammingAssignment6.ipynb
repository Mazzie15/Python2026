### Analyze which beverage category has:
- The most calories per drink?
- The most caffiene per drink?
- The most sugar per drink?
- The most proten per drink?

This will probably require you to write a for loop for each beverage in the beverage categories list.

In the loop you will need to get just the drinks that match each category from the dataframe and then sum up the calories, caffiene, sugar, and protein, for those drinks. Then divide by the total number of drinks in the category.

You will also need to keep track of the most calories, caffiene, sugar, and protein as you loop through to check the values. If your sum is greater then store both the amount and the category.

```python
most_cal = 0
most_caf = 0
most_sug = 0
most_pro = 0

most = {'Calories': None,
         'Caffeine': None,
         'Sugar': None,
         'Protein': None}


for drink in bev:
    mask_bev = DF['Beverage_category']==drink
    tmpDF = DF[mask_bev]
    num_drinks = sum(mask_bev)
    cals_per_drink = tmpDF['Calories'].sum()/num_drinks
    if cals_per_drink > most_cal:
        most_cal = cals_per_drink
        most['Calories'] = drink
    caf_per_drink = tmpDF['Caffeine (mg)'].sum()/num_drinks
    if caf_per_drink > most_caf:
        most_caf = caf_per_drink
        most['Caffeine'] = drink
    sug_per_drink = tmpDF['Sugars (g)'].sum()/num_drinks
    if sug_per_drink > most_sug:
        most_sug = sug_per_drink
        most['Sugar'] = drink
    pro_per_drink = tmpDF['Protein (g) '].sum()/num_drinks
    if pro_per_drink > most_pro:
        most_pro = pro_per_drink
        most['Protein'] = drink
    
most
```

### Sort the data by amt of sugar

Then make a bar plot of the top 30 most sugary drinks

```python
DF_sugary = DF.sort_values('Sugars (g)',ascending=False).head(30)

plt.figure(figsize=(16,8))
plt.title('Most Sugary Drinks: Starbucks')
x_values = DF_sugary['Drink']
y_values = DF_sugary['Sugars (g)']
plt.bar(x_values,y_values)
plt.xticks(rotation=90)
plt.grid()

plt.show()
```
