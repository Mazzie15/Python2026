### Dictionaries

- Create a dictionary that uses names as keys (start with your name) for the values you shoud use

- Add to your dictionary two more people and their information.

- Write code that asks a user if they want to be added to your contacts, and if so they are prompted to enter their age, birthday, and favoriate foods. These should be added to the dictionary.

````python
people = {
    'Mazzie': {
        'age': 21,
        'birthday': (8, 29),
        'fav_foods': ['beans','overnight oats','pasta','broccoli']
    },
    'Sydney': {
        'age': 23,
        'birthday': (1, 15),
        'fav_foods': ['pizza', 'chicken']
    },
    'Carmen': {
        'age': 21,
        'birthday': (9, 2),
        'fav_foods': ['ramen', 'salad','sushi']
    }
}
````

````python
for person in people:
    info = people[person]
    print(f"This is {person}!")
    print(f"Age: {info['age']}")
    print(f"Birthday: {info['birthday'][0]}/{info['birthday'][1]}")
    print("Favorite foods:")
    
    for food in info["fav_foods"]:
        print(f"  {food}")
````

````python
add_user = input("\nWould you like to be added to my contacts?").lower()

if add_user == "yes":
    
    name = input("Enter your name: ").capitalize()
    age = int(input("Enter your age: "))
    
    month = int(input("Enter your birth month: "))
    day = int(input("Enter your birth day: "))
    
    foods_input = input("Enter your favorite foods separated by commas: ")
    foods_list = foods_input.split(",")
    
    foods_list = [food.strip() for food in foods_list]
    
    people[name] = {
        "age": age,
        "birthday": (month, day),
        "fav_foods": foods_list
    }
    
    print("You have been added!")
else:
    print("Okay, maybe next time!")
````
