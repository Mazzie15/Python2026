### Problem 7

Loop through all the names in the list to find the shortest and longest name. Print those two names

```python
short = chr_names[0]
long = chr_names[0]

for name in chr_names:
    if len(name) < len(short):
        short = name
    elif len(name) > len(long):
        long = name

print(f"Shortest: {short}")
print(f"Longest: {long}")
```
