```python
def feeding_tracker(feedings, mode="total"):
    
    if mode == "none":
        print("No calculation performed. Returning original data.")
        return feedings
    elif mode == "total":
        totals = {}
        for pet in feedings:
            total = sum(feedings[pet])
            print(pet, "total food:", total)
            totals[pet] = total
        return totals
    elif mode == "avg":
        for pet in feedings:
            avg = sum(feedings[pet]) / len(feedings[pet])
            print(pet, "average feeding:", avg)
    elif mode == "most":
        for pet in feedings:
            print(pet, "largest feeding:", max(feedings[pet]))
    elif mode == "least":
        for pet in feedings:
            print(pet, "smallest feeding:", min(feedings[pet]))
    else:
        print("Error: invalid mode. Returning original data.")
        return feedings


pets = {
    "Timmy": [1.0, 1.2, 1.1],
    "Jack": [0.8, 1.0, 0.9],
    "Luna": [1.3, 1.4, 1.2]
}

feeding_tracker(pets)
feeding_tracker(pets, "avg")
feeding_tracker(pets, "most")
feeding_tracker(pets, "least")
feeding_tracker(pets, "none")
```
