```python
def add_item(item):
    file = open("bucket.txt", "a")
    file.write(item + "\n")
    file.close()
    print(f'"{item}" added to your bucket list!')


def remove_item(item):
    file = open("bucket.txt", "r")
    items = file.readlines()
    file.close()

    items = [i.strip() for i in items]

    if item in items:
        items.remove(item)
        file = open("bucket.txt", "w")
        for i in items:
            file.write(i + "\n")
        file.close()
        print(f'"{item}" removed from your bucket list.')
    else:
        print(f'"{item}" not found in your bucket list.')


def show_list():
    file = open("bucket.txt", "r")
    items = file.readlines()
    file.close()

    if len(items) == 0:
        print("Your bucket list is empty.")
    else:
        print("\nYour Bucket List:")
        for i in range(len(items)):
            print(f"{i+1}. {items[i].strip()}")


file = open("bucket.txt", "a")
file.close()

# User loop
while True:
    print("\nWhat would you like to do? (add, remove, show, quit)")
    choice = input("Enter choice: ").lower()

    if choice == "add":
        item = input("Enter item to add: ")
        add_item(item)

    elif choice == "remove":
        item = input("Enter item to remove: ")
        remove_item(item)

    elif choice == "show":
        show_list()

    elif choice == "quit":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Please try again.")
```
